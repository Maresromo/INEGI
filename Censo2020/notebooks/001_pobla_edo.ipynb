```python
f = '../../ENIGH/data/raw/catun_municipio/AGEEML_20258131423976_utf.csv'
ubica_geo = pd.read_csv(f, encoding="utf-8"
                       # usecols = ['folioviv','aire_acond','entidad','factor']
                      )
ubica_geo.head()
```

```python
nombre_edo = "Campeche"
ubica_edo = ubica_geo[ubica_geo["NOM_ENT"].str.contains(nombre_edo, case=False, na=False)]
codigo_edo = ubica_edo['CVE_ENT'].mean()
```

```python
nombre_mun = "San Francisco de Campeche"
ubica_mun = ubica_geo[ubica_geo["NOM_CAB"].str.contains(nombre_mun, case=False, na=False)]
codigo_mun = ubica_mun['CVE_MUN'].mean()
```

```python
import pandas as pd

f_viv = '../data/Viviendas00.CSV'
f_per = '../data/Personas00.CSV'

# Variables de filtrado
# codigo_mun = ...
# codigo_edo = ...

# Leer por partes y filtrar
vivienda = pd.concat(
    (chunk[(chunk['MUN'] == codigo_mun) & (chunk['ENT'] == codigo_edo)]
     for chunk in pd.read_csv(f_viv, chunksize=100000)),
    ignore_index=True
)

poblacion = pd.concat(
    (chunk[(chunk['MUN'] == codigo_mun) & (chunk['ENT'] == codigo_edo)]
     for chunk in pd.read_csv(f_per, chunksize=100000)),
    ignore_index=True
)
```

```python
solo_en_poblacion = list(set(poblacion.columns) - set(vivienda.columns))
solo_en_vivienda = list(set(vivienda.columns) - set(poblacion.columns))

print("Elementos solo en poblacion:", len(solo_en_poblacion))
print("Elementos solo en vivienda:", len(solo_en_vivienda))
print("Elementos en poblacion en total:", len(poblacion.columns))
print("Elementos en vivienda en total:", len(vivienda.columns))
```
