```python
f = '../../ENIGH/data/raw/catun_municipio/AGEEML_20258131423976_utf.csv'
ubica_geo = pd.read_csv(f, encoding="utf-8"
                       # usecols = ['folioviv','aire_acond','entidad','factor']
                      )
ubica_geo
```

```python
nombre_edo = "Oaxaca"
ubica_edo = ubica_geo[ubica_geo["NOM_ENT"].str.contains(nombre_edo, case=False, na=False)]
codigo = ubica_edo['CVE_ENT'].mean()
```

```python
import pandas as pd

f = '../data/Viviendas00.CSV'
# codigo = 20  # Clave del estado que quieres filtrar

# Leer el CSV por partes para evitar cargar todo en memoria
chunks = pd.read_csv(f, chunksize=100000)

# Filtrar solo las filas donde ENT == codigo
viv = pd.concat([chunk[chunk['ENT'] == codigo] for chunk in chunks])
```

```python
viv.columns
```

```python
viv['FACTOR'].sum()
```

Para Oaxaca en la ENIGH 2024 hay 1138552 viviendas en Oaxaca

```python
viv['MUN'].nunique(dropna=True)
```

```python
suma_factor_por_mun = viv.groupby('MUN', as_index=False)['FACTOR'].sum()
suma_factor_por_mun.rename(columns={'FACTOR': 'suma_factor'}, inplace=True)
suma_factor_por_mun
```

```python
suma_factor_por_mun['suma_factor'].min()
```
