```python
from osgeo import gdal

# Ruta a tu archivo GeoPDF
geopdf_path = "../data/Oaxaca.pdf"

# Abrir el GeoPDF
dataset = gdal.OpenEx(geopdf_path, gdal.OF_VECTOR)

# Listar capas vectoriales disponibles
for i in range(dataset.GetLayerCount()):
    layer = dataset.GetLayerByIndex(i)
    print(f"Capa {i}: {layer.GetName()}")
```

```python
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
# Ruta al shapefile de municipios
path = '../data/702825218874_s/conjunto de datos/20m.shp'
# path = "../data/oaxaca/20m.shp"

# Leer shapefile
gdf = gpd.read_file(path)

# Ver columnas disponibles
print(gdf.columns)

# Mostrar las primeras filas
print(gdf.head())

# Graficar el mapa de municipios
gdf.plot(figsize=(10, 10), edgecolor='black')
plt.title("Municipios de Oaxaca (INEGI)")
plt.show()
```
